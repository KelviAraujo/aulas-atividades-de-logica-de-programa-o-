```python
numero = []
for i in range(7):
    numero.append(int(input("numero: ")))

maior = numero[0]
menor = numero[0]
soma = 0

for n in numero:
    soma = soma = n
    if n > maior:
        maior = n
    if n < menor:
        menor = n 
        
print("soma: ", soma)
print("maior:", maior)
print("menor:", menor)
```

```python
todos = []
pares = []
impares = []
for i in range(10):
    n = int(input("Digite: "))
    todos.append(n)
    if n % 2 == 0:
        pares.append(n)
    else:
        impares.append(n)
print("Pares:", pares)
print("Ímpares:", impares)
```

```python
esportes = []
while True:
    e = input("Esporte (fim para sair): ")
    if e == "fim":
        break
    esportes.append(e)
esportes.sort()
print(len(esportes), "esportes:", esportes)
```

```python
medições = []
for i in range(8):
    medições.append(float(input("Valor: ")))

soma = 0
for m in medições:
    soma = soma + m
media = soma / 8

for m in medições:
    if m > media:
        print("Acima da média:", m)
```
